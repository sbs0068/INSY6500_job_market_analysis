Missing data points found in relevant columns-`salary_min`, `salary_max`, `salary_mid`-as well as in columns describing contract types and locations. Comparing missing entries to the `salary_predicted` columns, it is not dependant on whether the other salary columns are filled or not.

An important note is that the number of `NaN` entries in `salary_min` and `salary_mid` are equal, while `salary_max` has 9 more missing entries. This tells us that `salary_mid` is only reliant on `salary_mid` to calculate a score. We may need to ignore the 9 extra `NaN` entries in `salary_max`

There are 14 `id` entries that are duplicate entries so these must be removed from the dataframe.


```python
missing_salary_mask = df['salary_min'].isna() & df['salary_max'].isna()
check_mask = missing_salary_mask & (df['salary_is_predicted'] == 0)

# Count how many rows meet this condition
count_check = check_mask.sum()
print(f"Number of rows with missing salaries AND salary_is_predicted == 0: {count_check}")
total_missing = missing_salary_mask.sum()
print(f"Total missing salaries: {total_missing}")
print(f"Percentage of missing salaries with salary_is_predicted== 0: {count_check / total_missing:.2%}")
```

There is a 100% match on missing salaries where they are not predicted. This means that the missing salary values are as expected and there is no need for correction to these values.

## Step 2: Cleaning Decisions
This step resolves problems discovered in Step 1

Right now I think we should remove any duplicated rows, isolate the rows where there are significant outliers but not all. Just the few in the millions as they are too big to ignore. I feel the others could be relevant data points and interesting to see.  

```python
# Remove Duplicates for further analysis and create a copy of the df
df_no_duplicates = df[~df['id'].duplicated(keep=False)].copy()


```

The first step in cleaning the data was to remove the duplicated rows. From inspecting the time stamps of these duplicates it was decided that they were simply misclicks or errors by the internet scrapper used to compile a large chunk of this dataset. 



```python
# Build the mask 
mask = df_no_duplicates['salary_max'].isna() & df_no_duplicates['salary_min'].notna()

# Fillmax salaries with min
df_no_duplicates.loc[mask, 'salary_max'] = df_no_duplicates.loc[mask, 'salary_min']


```

There were several rows where the `salary_max` was missing but not the `salary_min`. Since in these cases they should be treated as the same value we elected to use the fill method to fill in these missing max salary values. 


```python
# Keep only rows where salary_max is between 1000 and 1,000,000
df_filtered = df[(df['salary_max'] >= 1000) & (df['salary_max'] <= 1_000_000)]
df_filtered
```

Due to the large effect they have on the dataset, we decided to remove salaries that were over 1 million dollars in compensation. It was also decided to remove salary information under $1,000 as these were most likely hourly pay rates that were detected by the sweeper.

```python
df_filtered.to_pickle("../data/cleaned_data.pkl")
```
